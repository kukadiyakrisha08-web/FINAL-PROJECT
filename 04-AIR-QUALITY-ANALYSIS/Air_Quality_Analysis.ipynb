```python
# Time માં . ને : માં convert કરો
df["Time"] = df["Time"].astype(str).str.replace(".", ":", regex=False)

# Date અને Time combine કરીને DateTime બનાવો
df["DateTime"] = pd.to_datetime(
    df["Date"].astype(str) + " " + df["Time"],
    format="%d/%m/%Y %H:%M:%S",
    errors="coerce"
)
```

```python
df.head()
```

```python
df["CO(GT)"] = pd.to_numeric(
    df["CO(GT)"].astype(str).str.replace(",", ".", regex=False),
    errors="coerce"
)
```

```python
print(df["CO(GT)"].dtype)
```

```python
plt.figure(figsize=(12, 5))

plt.plot(df.index, df["CO(GT)"])

plt.title("CO Air Pollution Trend Over Time")
plt.xlabel("Date and Time")
plt.ylabel("CO Concentration")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
df["NOx(GT)"] = pd.to_numeric(
    df["NOx(GT)"].astype(str).str.replace(",", ".", regex=False),
    errors="coerce"
)

df["NO2(GT)"] = pd.to_numeric(
    df["NO2(GT)"].astype(str).str.replace(",", ".", regex=False),
    errors="coerce"
)
```

```python
plt.figure(figsize=(12, 5))

plt.plot(df.index, df["NOx(GT)"], label="NOx")
plt.plot(df.index, df["NO2(GT)"], label="NO2")

plt.title("NOx and NO2 Pollution Comparison")
plt.xlabel("Date and Time")
plt.ylabel("Concentration")

plt.legend()
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
