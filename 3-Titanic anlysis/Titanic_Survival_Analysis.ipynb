```python
sns.countplot(x='Survived', data=df)

plt.title('Survival Count')
plt.xlabel('Survived (0 = No, 1 = Yes)')
plt.ylabel('Number of Passengers')

plt.show()
```

```python
df['Survived'].value_counts()
```

```python
survival_rate = df['Survived'].mean() * 100

print(f"Overall Survival Rate: {survival_rate:.2f}%")
```

```python
sns.countplot(x='Sex', hue='Survived', data=df)

plt.title('Survival by Gender')
plt.xlabel('Gender')
plt.ylabel('Number of Passengers')

plt.show()
```

```python
gender_survival = df.groupby('Sex')['Survived'].mean() * 100

print(gender_survival)
```

```python
df.groupby('Sex')['Survived'].mean()
```

```python
sns.countplot(x='Pclass', hue='Survived', data=df)

plt.title('Survival by Passenger Class')
plt.xlabel('Passenger Class')
plt.ylabel('Number of Passengers')

plt.show()
```

```python
class_survival = df.groupby('Pclass')['Survived'].mean() * 100

print(class_survival)
```

```python
class_survival.plot(kind='bar')

plt.title('Survival Rate by Passenger Class')
plt.xlabel('Passenger Class')
plt.ylabel('Survival Rate (%)')

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(df['Age'], bins=30, kde=True)

plt.title('Age Distribution of Titanic Passengers')
plt.xlabel('Age')
plt.ylabel('Number of Passengers')

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(x='Survived', y='Age', data=df)

plt.title('Age Distribution by Survival')
plt.xlabel('Survived (0 = No, 1 = Yes)')
plt.ylabel('Age')

plt.show()
```

```python
bins = [0, 12, 18, 30, 50, 100]

labels = ['Child', 'Teenager', 'Young Adult', 'Adult', 'Senior']

df['AgeGroup'] = pd.cut(df['Age'], bins=bins, labels=labels)
```

```python
df[['Age', 'AgeGroup']].head(10)
```

```python
age_survival = df.groupby('AgeGroup', observed=False)['Survived'].mean() * 100

print(age_survival)
```

```python
age_survival.plot(kind='bar')

plt.title('Survival Rate by Age Group')
plt.xlabel('Age Group')
plt.ylabel('Survival Rate (%)')

plt.xticks(rotation=45)

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(df['Fare'], bins=30, kde=True)

plt.title('Fare Distribution of Titanic Passengers')
plt.xlabel('Fare')
plt.ylabel('Number of Passengers')

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(x='Survived', y='Fare', data=df)

plt.title('Fare Distribution by Survival')
plt.xlabel('Survived (0 = No, 1 = Yes)')
plt.ylabel('Fare')

plt.show()
```

```python
sns.countplot(x='Embarked', hue='Survived', data=df)

plt.title('Survival by Embarkation Port')
plt.xlabel('Embarkation Port')
plt.ylabel('Number of Passengers')

plt.show()
```

```python
embarked_survival = df.groupby('Embarked')['Survived'].mean() * 100

print(embarked_survival)
```

```python
embarked_survival.plot(kind='bar')

plt.title('Survival Rate by Embarkation Port')
plt.xlabel('Embarkation Port')
plt.ylabel('Survival Rate (%)')

plt.show()
```

```python
df['FamilySize'] = df['SibSp'] + df['Parch'] + 1

df[['SibSp', 'Parch', 'FamilySize']].head(10)
```

```python
family_survival = df.groupby('FamilySize')['Survived'].mean() * 100

print(family_survival)
```

```python
plt.figure(figsize=(10,5))

family_survival.plot(kind='bar')

plt.title('Survival Rate by Family Size')
plt.xlabel('Family Size')
plt.ylabel('Survival Rate (%)')

plt.show()
```

```python
plt.figure(figsize=(10,7))

numeric_df = df.select_dtypes(include=np.number)

sns.heatmap(
    numeric_df.corr(),
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Correlation Heatmap')

plt.show()
```
