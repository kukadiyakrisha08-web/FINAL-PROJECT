```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
import glob

file_path = glob.glob('/content/*.csv')[0]

df = pd.read_csv(file_path)

df.head()
```

```python
print("Rows and Columns:", df.shape)

print("\nColumn Names:")
print(df.columns)

print("\nDataset Information:")
df.info()
```

```python
df.nunique()
```

```python
top_10_happiest = df.sort_values(by="Score", ascending=False).head(10)

print("Top 10 Happiest Countries:")
print(top_10_happiest[["Country or region", "Score"]])
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))

plt.barh(
    top_10_happiest["Country or region"],
    top_10_happiest["Score"]
)

plt.xlabel("Happiness Score")
plt.ylabel("Country")
plt.title("Top 10 Happiest Countries")

plt.gca().invert_yaxis()

plt.show()
```

```python
bottom_10_happiest = df.sort_values(by="Score", ascending=True).head(10)

print("Bottom 10 Least Happy Countries:")
print(bottom_10_happiest[["Country or region", "Score"]])
```

```python
plt.figure(figsize=(10, 6))

plt.barh(
    bottom_10_happiest["Country or region"],
    bottom_10_happiest["Score"]
)

plt.xlabel("Happiness Score")
plt.ylabel("Country")
plt.title("Bottom 10 Least Happy Countries")

plt.show()
```

```python
plt.figure(figsize=(10, 6))

plt.scatter(
    df["GDP per capita"],
    df["Score"]
)

plt.xlabel("GDP per Capita")
plt.ylabel("Happiness Score")
plt.title("GDP per Capita vs Happiness Score")

plt.show()
```

```python
gdp_correlation = df["GDP per capita"].corr(df["Score"])

print("Correlation between GDP per Capita and Happiness Score:", gdp_correlation)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

correlation_matrix = df.select_dtypes(include="number").corr()

plt.figure(figsize=(12, 8))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap of Happiness Factors")
plt.show()
```

```python
score_correlations = df.select_dtypes(include="number").corr()["Score"].sort_values(ascending=False)

print("Correlation with Happiness Score:")
print(score_correlations)
```

```python
score_correlations.drop("Score").sort_values().plot(
    kind="barh",
    figsize=(10, 6)
)

plt.xlabel("Correlation with Happiness Score")
plt.ylabel("Factor")
plt.title("Factors Affecting Happiness Score")
plt.show()
```

```python
correlations = df.select_dtypes(include="number").corr()["Score"]

correlations = correlations.drop("Score").sort_values(
    ascending=False
)

print("Factors most correlated with Happiness Score:")
print(correlations)
```

```python
plt.figure(figsize=(10, 6))

correlations.sort_values().plot(kind="barh")

plt.xlabel("Correlation")
plt.ylabel("Factors")
plt.title("Correlation of Factors with Happiness Score")

plt.show()
```
